Изучите приведенную ниже простейшую программу для GPU.

Изучите спецификатор __global__ , и конструкцию запуска ядра <<<...>>> .
Не забудьте, что расширение файла должно быть .cu .

```c++
#include <stdio.h>

__global__ void HelloKernel()
{
   printf("Hello world!");
}

int main()
{
   HelloKernel<<<1, 1>>>();
   return 0;
}
```

Соберите и запустите программу:
```bash
$ nvcc main.cu -o main
$ ./main
```

````python
%%writefile main.cu
#include <stdio.h>

__global__ void HelloKernel()
{
   printf("Hello world!");
}

int main()
{
   HelloKernel<<<2, 1>>>();
   cudaDeviceSynchronize();
   return 0;
}
````

### Что делает `__global__`

Ключевое слово `__global__` в CUDA говорит компилятору:

> «Эта функция — **ядро CUDA**, и её нужно выполнять **на GPU**, а вызывать — **с CPU**».

То есть:

* `__global__` → функция запускается **на видеокарте (GPU)**;
* `<<<...>>>` → синтаксис, который говорит:
  **сколько блоков** и **по сколько потоков** в каждом блоке запустить.

### Почему `printf("Hello world!")` не выводится в Colab

Всё дело не в `__global__`, а в **ограничениях Colab**:

* Colab **не перенаправляет вывод GPU** (`stdout` устройства) в терминал.
  Поэтому `printf` внутри GPU просто «теряется».
* Программа реально **выполняется**, просто ты не видишь результат.

То есть:

```cpp
__global__ void HelloKernel() {
    printf("Hello world!");
}
```

работает — но ты не увидишь текст.

---

### Итак:

| Слово        | Где выполняется | Где вызывается | Для чего                                    |
| ------------ | --------------- | -------------- | ------------------------------------------- |
| `__global__` | GPU             | CPU            | Функция-ядро, запускается на видеокарте     |
| `__device__` | GPU             | GPU            | Вспомогательная функция, только внутри ядра |
| `__host__`   | CPU             | CPU            | Обычная функция CPU (по умолчанию)          |
