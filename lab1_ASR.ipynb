```python
from pydub import AudioSegment

def cut_audio(file, start_min, start_sec, end_min, end_sec, output):
    audio = AudioSegment.from_file(file)
    start_time = (start_min * 60 + start_sec) * 1000
    end_time = (end_min * 60 + end_sec) * 1000
    fragment = audio[start_time:end_time]
    fragment.export(output, format="wav")
    print(f"Сохранено: {output}")
```

```python

cut_audio("speaker1.mp3", 2, 58, 3, 18, "speaker_Guber.wav")
cut_audio("speaker2-3.mp3", 10, 00, 10, 20, "speaker_Burunov.wav")
cut_audio("speaker2-3.mp3", 1, 32, 1, 52, "speaker_Basta.wav")
cut_audio("speaker4.mp3", 0, 8, 0, 23, "speaker_Dora.wav")
cut_audio("speaker5.mp3", 2, 58, 3, 18, "speaker_Aranova.wav")
```
