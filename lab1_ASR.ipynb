```python
from pydub import AudioSegment

def cut_audio(file, start_min, start_sec, end_min, end_sec, output):
    audio = AudioSegment.from_file(file)
    start_time = (start_min * 60 + start_sec) * 1000
    end_time = (end_min * 60 + end_sec) * 1000
    fragment = audio[start_time:end_time]
    fragment.export(output, format="wav")
    print(f"Сохранено: {output}")
```

```python

cut_audio("speaker1.mp3", 2, 58, 3, 18, "speaker_Guber.wav")
cut_audio("speaker2-3.mp3", 10, 00, 10, 20, "speaker_Burunov.wav")
cut_audio("speaker2-3.mp3", 1, 32, 1, 52, "speaker_Basta.wav")
cut_audio("speaker4.mp3", 0, 8, 0, 23, "speaker_Dora.wav")
cut_audio("speaker5.mp3", 2, 58, 3, 18, "speaker_Aranova.wav")
```

Загружаем модель

Прогоним распознавание для всех файлов. Результат – распознанные тексты

```python
import os
import whisperx

model = whisperx.load_model("large", device="cpu", compute_type="int8")

results = {}
for file in os.listdir():
    if file.endswith(".wav"):
        audio = whisperx.load_audio(file)
        result = model.transcribe(audio, batch_size=8)

        # собираем весь текст из сегментов
        text = " ".join([seg["text"] for seg in result["segments"]])
        results[file] = text

        print(file, "→", text)

```

Подсчёт метрик WER и CER

```python
from jiwer import wer, cer

# Импортируем метрики
for file, whisper_text in results.items():
    my_text = manual[file]
    print(f"{file}:")
    print("  WER:", wer(my_text, whisper_text))
    print("  CER:", cer(my_text, whisper_text))
    print()
```

WER (Word Error Rate) — процент ошибок по словам.
CER (Character Error Rate) — процент ошибок по символам.

Чем ближе к 0 → тем лучше распознавание
